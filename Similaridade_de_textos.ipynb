```python
similaridade = ['abacate', 'abacaxi','abobora','abobrinha','ananás', 'maçã', 'mamão', 'manga', 'melancia','melão', 'mexerica', 'morango']

entrada = input('Digite uma fruta:   ')
print(entrada)

def getDigrama(texto):
  digramas = []
  for palavra in range(len(texto)-1):
    digramas.append(texto[palavra:palavra+2])
  return digramas

def getUnicos(texto):
  return set(texto)

def getCompartilhados(texto1, texto2):
  compartilhados = []
  for i in texto1:
    for j in texto2:
      if i in j:
        compartilhados.append(i)
        break
  return compartilhados

def getMetrica(a, b, c):
  return (2 * c / (a + b))
```

```python
digrama_entrada = getDigrama(entrada)
unicos_entrada = getUnicos(digrama_entrada)
```

```python
resultado = []

for fruta in similaridade:
    digrama_fruta = getDigrama(fruta)
    unicos_fruta = getUnicos(digrama_fruta)
    compartilhados = getCompartilhados(unicos_entrada, unicos_fruta)
    metricas = getMetrica(len(unicos_entrada), len(unicos_fruta), len(compartilhados))
    resultado.append({
        "Fruta": fruta,
        "Digramas": digrama_fruta,
        "Únicos": digrama_entrada,
        "Compartilhados": compartilhados,
        "Similaridade": metricas
    })

import pandas as pd

df = pd.DataFrame(resultado)
df = df.sort_values(by='Similaridade', ascending=False)
df.head(4)
```
